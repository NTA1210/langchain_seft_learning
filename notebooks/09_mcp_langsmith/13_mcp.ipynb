# 13 — Model Context Protocol (MCP)

## Learning requirements
Phân biệt rõ:

**Tool calling** = model/application mechanism để model yêu cầu dùng tool.

**MCP** = protocol chuẩn để expose/discover tools, resources và prompts giữa AI client và external capability server.

Architecture:

```text
Agent
  -> MCP client
      |- GitHub MCP server
      |- Filesystem MCP server
      `- Custom project MCP server
```

## Core concepts

- MCP server
- MCP client
- tools
- resources
- prompts
- stdio / network transports
- session lifecycle
- trust boundary
- auth/permissions

Không tự động tin MCP tool description/output. Third-party MCP server là external trust boundary.

````python
# Minimal conceptual MCP server.
# Exact server runner/transport options may evolve; inspect current `mcp` package docs.
from mcp.server.fastmcp import FastMCP

mcp = FastMCP("project-tools")

@mcp.tool()
def get_project(project_id: str) -> dict:
    """Return safe project metadata for one exact project ID."""
    demo = {"lcsp": {"name": "LCSP", "status": "active"}}
    return demo.get(project_id, {"found": False})

@mcp.tool()
def search_documents(query: str) -> list[dict]:
    """Search demo project documents. Replace with real retrieval in the exercise."""
    return [{"source": "demo.md", "content": f"Demo result for {query}"}]

# Run from a normal .py server entrypoint rather than inside Jupyter when needed.
# mcp.run()
````
