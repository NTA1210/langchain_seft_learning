## Evaluation layers

### Offline evaluation
Dataset cố định -> Agent v1/v2 -> Evaluators -> compare.

### Online evaluation
Production traces -> sampled evaluators/feedback -> monitoring.

### Agent evaluation
Không chỉ final answer. Đánh giá:
- routing;
- tool selection;
- tool arguments;
- retrieval quality;
- trajectory;
- final response;
- safety/permission compliance.

````python
# Start with a local evaluation dataset independent from LangSmith APIs.
eval_cases = [
    {
        "input": "Calculate 12 * 9",
        "expected_tool": "multiply",
        "expected_answer_contains": "108",
    },
    {
        "input": "Say hello",
        "expected_tool": None,
        "expected_answer_contains": "hello",
    },
]

def tool_selection_score(actual_tool: str | None, expected_tool: str | None) -> int:
    return int(actual_tool == expected_tool)

print(tool_selection_score("multiply", "multiply"))
````

## Required output — `artifacts/evaluation-results.md`

Dataset tối thiểu 20–50 cases.

Report ít nhất:
- task success;
- correct tool selection;
- invalid tool args;
- retrieval/citation score nếu có RAG;
- safety violations;
- latency;
- tokens/cost;
- v1 vs v2 comparison.

Ví dụ format:
```text
correct tool selection: 82% -> 94%
task success:           76% -> 89%
average latency:        4.8s -> 4.2s
```
